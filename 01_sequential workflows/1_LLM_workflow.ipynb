```python
'''
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

llm=HuggingFaceEndpoint(
    model="Qwen/Qwen2.5-7B-Instruct",
    temperature=0.7,
    max_new_tokens=200
)
model=ChatHuggingFace(llm=llm)    # this one is not working because we dont have credits in our account. So we will use Ollama model instead.
'''
from langchain_ollama import ChatOllama

model = ChatOllama(
    model="qwen2.5:3b",
    temperature=0.7
)

```

```python
#create a state
from typing import TypedDict

class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    #extract the question from the state
    question=state['question']
    #create a prompt for the model
    prompt=f"Answer the following question: {question}"
    #invoke the model with the prompt
    answer=model.invoke(prompt).content
    #update the ans in the state
    state['answer']=answer
    return state
```

```python
#create our graph
from langgraph.graph import StateGraph, START, END
graph=StateGraph(LLMState)

#add nodes
graph.add_node("llm_qa",llm_qa)
#add edges
graph.add_edge(START,"llm_qa")
graph.add_edge("llm_qa",END)
#compile the graph
workflow=graph.compile()
```

```python
#execute the graph
result=workflow.invoke({"question": "how far is moon from earth?"})
print(result)  # Output: {'question': 'What is the capital of France?', '
```
