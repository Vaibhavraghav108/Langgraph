```python
from langgraph.graph import StateGraph,START,END
from langchain_ollama import ChatOllama
from typing import TypedDict
from dotenv import load_dotenv

load_dotenv()

model = ChatOllama(
    model="qwen3:1.7b",
    temperature=0.7
)
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:
    #fetch title
    title=state['title']
    #call llm gen outline
    promt=f"Create a detailed outline for a blog post with the title: '{title}'."
    outline=model.invoke(promt).content
    state['outline'] = outline
    return state
```

```python
def create_blog(state: BlogState)->BlogState:
    title=state['title']
    outline=state['outline']
    promt=f"Create a detailed blog with the title: '{title}' and the following outline: '{outline}'."
    content=model.invoke(promt).content
    state['content'] = content
    return state
```

```python
graph=StateGraph(BlogState)

#add nodes
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)

#add edges
graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog',END)

workflow=graph.compile()

```

```python
result=workflow.invoke({"title": "Rise of AI in India"})
print(result)  # Output: {'title': 'The Future of AI in Healthcare', 'outline': '...', 'content': '...'}
```
